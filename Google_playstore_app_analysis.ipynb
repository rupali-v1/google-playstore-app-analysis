The data in size column is present in kilobytes as well as and we also have varies with device where we will impute NaN. So we will convert evrything to megabytes.

````python
def convert_size_to_mb(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().upper()

    if value == 'VARIES WITH DEVICE':
        return np.nan

    if value.endswith('M'):
        return float(value[:-1])

    if value.endswith('K'):
        return float(value[:-1]) / 1000

    return np.nan
````

````python
apps_clean['Size'] = apps_clean['Size'].apply(convert_size_to_mb)
````
